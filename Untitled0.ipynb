```python
ko = yf.download("KO", start="2019-01-01")
returns = ko["Close"].pct_change()
print("annualized return:", float(returns.mean() * 252))
print("annualized volatility:", float(returns.std() * 252 ** 0.5))
ko["Close"].plot(title="KO since 2019")
```

```python
nvda = yf.download("NVDA", start="2019-01-01")
r = nvda["Close"].pct_change()
print("annualized return:", float(r.mean() * 252))
print("annualized volatility:", float(r.std() * 252 ** 0.5))
```

Risk-adjusted returns: KO vs NVDA

I compared Coca-Cola (KO) and NVIDIA (NVDA) using daily price data since 2019. Before running the numbers, I predicted NVDA would have higher returns and lower volatility than KO. I was wrong about volatility: NVDA returned ~68% annually with ~51% volatility, versus ~13% / ~20% for KO. NVDA's valuation is driven by expectations about AI, which makes the price swing much more.

Surprisingly, NVDA still wins on a risk-adjusted basis: return/volatility of 1.34 vs 0.66 (a rough cousin of the Sharpe ratio, ignoring the risk-free rate).

Limitations: returns exclude dividends (understating KO), and the comparison suffers from hindsight bias — I picked NVDA already knowing the AI boom happened. Past volatility tends to persist; past returns don't.
